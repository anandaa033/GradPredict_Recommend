```python
import pandas as pd
import numpy as np

from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import KFold, cross_val_score
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error
)

# =====================================================
# LOAD DATA
# =====================================================

df = pd.read_csv('dataSet/DataSet.csv')

# แปลง categorical -> numerical
df = pd.get_dummies(df)

# =====================================================
# DEFINE X AND y
# =====================================================

target = 'ระยะเวลาในเดือน'

X = df.drop(columns=[target])
y = df[target]

# =====================================================
# CREATE MODEL
# =====================================================

model = DecisionTreeRegressor(
    random_state=38,
    max_depth=5
)

# =====================================================
# K-FOLD CROSS VALIDATION
# =====================================================

kfold = KFold(
    n_splits=5,
    shuffle=True,
    random_state=38
)

# ---------- R² ----------
r2_scores = cross_val_score(
    model,
    X,
    y,
    cv=kfold,
    scoring='r2'
)

# ---------- MSE ----------
mse_scores = cross_val_score(
    model,
    X,
    y,
    cv=kfold,
    scoring='neg_mean_squared_error'
)

mse_scores = -mse_scores

# ---------- RMSE ----------
rmse_scores = np.sqrt(mse_scores)

# ---------- MAE ----------
mae_scores = cross_val_score(
    model,
    X,
    y,
    cv=kfold,
    scoring='neg_mean_absolute_error'
)

mae_scores = -mae_scores

# =====================================================
# SHOW RESULTS
# =====================================================

print("\n========== K-FOLD RESULT ==========\n")

print("R² ของแต่ละ Fold:")
print(r2_scores)

print(f"\nAverage R²: {r2_scores.mean()*100:.2f}%")

print("\nMSE ของแต่ละ Fold:")
print(mse_scores)

print(f"\nAverage MSE: {mse_scores.mean():.2f}")

print("\nRMSE ของแต่ละ Fold:")
print(rmse_scores)

print(f"\nAverage RMSE: {rmse_scores.mean():.2f}")

print("\nMAE ของแต่ละ Fold:")
print(mae_scores)

print(f"\nAverage MAE: {mae_scores.mean():.2f}")

# =====================================================
# TRAIN WITH ALL DATA
# =====================================================

model.fit(X, y)

# =====================================================
# CUSTOM INPUT
# =====================================================

data_test = {
    '2.นักศึกษาสังกัดหลักสูตร': ['ปร.ด.สาขาวิชาเทคโนโลยียาง'],
    '3. เพศ': ['หญิง'],
    '4. อายุ (ปี)': [25],
    '6. สถานภาพสมรส (ปัจจุบัน)': ['โสด'],
    '7. ท่านมีความต้องการสำเร็จการศึกษาตามระยะเวลาที่หลักสูตรกำหนดหรือไม่': ['ต้องการ'],
    '8. ท่านทราบและเข้าใจถึงหลักเกณฑ์และวิธีการสำเร็จการศึกษาระดับบัณฑิตศึกษาหรือไม่': ['ทราบ']
}

# =====================================================
# PREPARE TEST DATA
# =====================================================

df_test = pd.DataFrame(data_test)

df_test = pd.get_dummies(df_test)

df_test = df_test.reindex(columns=X.columns, fill_value=0)

# =====================================================
# PREDICT
# =====================================================

y_pred = model.predict(df_test)

predicted_months = y_pred[0]
predicted_years = predicted_months / 12

print("\n========== PREDICTION ==========\n")

print(f"คาดว่าจะจบภายในประมาณ {predicted_years:.1f} ปี")

avg_month = y.mean()

percent = (predicted_months / avg_month) * 100

print(f"เปอร์เซ็นต์เทียบค่าเฉลี่ย: {percent:.2f}%")
```
